```python
from ctransformers import AutoModelForCausalLM
import time

print("=== CPU 모드 (gpu_layers=0) ===")
start = time.time()
model_cpu = AutoModelForCausalLM.from_pretrained(
    "TheBloke/Llama-2-7B-Chat-GGUF",
    model_file="llama-2-7b-chat.Q4_K_M.gguf",
    model_type="llama",
    gpu_layers=0,
    context_length=512,
)
print(f"로드: {time.time()-start:.1f}초")

prompt = "Hello, what is 2+2?"
start = time.time()
out = model_cpu(prompt, max_new_tokens=64)
cpu_time = time.time() - start
print(f"응답: {out}")
print(f"생성 시간 (CPU): {cpu_time:.1f}초\n")
del model_cpu
```

```python
print("=== GPU 모드 (gpu_layers=50) ===")
start = time.time()
model_gpu = AutoModelForCausalLM.from_pretrained(
    "TheBloke/Llama-2-7B-Chat-GGUF",
    model_file="llama-2-7b-chat.Q4_K_M.gguf",
    model_type="llama",
    gpu_layers=50,
    context_length=512,
)
print(f"로드: {time.time()-start:.1f}초")

start = time.time()
out = model_gpu(prompt, max_new_tokens=64)
gpu_time = time.time() - start
print(f"응답: {out}")
print(f"생성 시간 (GPU): {gpu_time:.1f}초\n")
del model_gpu
```

```python
print("=== 속도 비교 ===")
print(f"CPU: {cpu_time:.1f}초")
print(f"GPU: {gpu_time:.1f}초")
print(f"GPU 가속: {cpu_time/gpu_time:.1f}배 빠름")
```
